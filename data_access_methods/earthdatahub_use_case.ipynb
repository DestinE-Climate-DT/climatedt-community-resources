#### Earth Data Hub simplified use case 

This Jupyter notebook provides a simplified example of accessing the data with Earth Data Hub and plotting the data. The code use in this notebook could also be found [here](https://earthdatahub.destine.eu/tutorials/tutorial-era5-single-levels-climatological-analysis-of-temperature-in-germany).  

You may find additional examples at https://earthdatahub.destine.eu/tutorials, where you could download the jupyter notebook and test it in this project. Alternatively, you may access Insula Code service to test the examples on the Web. 

```python
import xarray as xr

EDH_KEY = "your_personal_access_token"
#"edh_key_e5nBDG1UrDeNdtx4Cboj-E6rZII1uk2GIzDgWaW-aRo"

url = f"https://edh:{EDH_KEY}@api.earthdatahub.destine.eu/climate-dt-2/IFS-NEMO-SSP3-7.0-sfc-hourly-standard-v0.zarr"

ds = xr.open_dataset(
    url,
    chunks={},
    engine="zarr",
    zarr_format=3,
    storage_options={
    "client_kwargs": {
        "trust_env":True
    }
}
)
ds
```

```python
xr.set_options(keep_attrs=True)

t2m = ds.t2m - 273.15
t2m.attrs["units"] = "°C"
t2m
```

```python
germany = {'latitude': slice(46, 56), 'longitude': slice(5, 16)}
t2m_germany = t2m.sel(**germany)
t2m_germany
```
